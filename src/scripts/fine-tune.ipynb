```python
import transformers
from transformers import AutoModelForCausalLM,Pipeline,AutoTokenizer,Pipeline
from peft import LoraConfig,get_peft_model
from datasets import load_dataset
import transformers
from peft import  prepare_model_for_kbit_training
from accelerate import Accelerator
import torch
```

```python
model_name = "mistralai/Mistral-7B-Instruct-v0.2"
model = AutoModelForCausalLM.from_pretrained(model_name,
                                             device_map="auto", # automatically figures out how to best use CPU + GPU for loading model
                                             trust_remote_code=False, # prevents running custom model files on your machine
                                              revision="main") # which version of model to use in repo
```

```python
tokenizer=AutoTokenizer.from_pretrained(task='text-generation',pretrained_model_name_or_path=model_name,use_fast=True)
```

```python
from transformers import pipeline

pipeline = pipeline(task="text-generation", model=model, tokenizer=tokenizer)
pipeline("the secret to baking a really good cake is ")
[{'generated_text': 'the secret to baking a really good cake is 1. the right ingredients 2. the'}]
```
